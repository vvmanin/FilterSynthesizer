```python
import pathlib, re
HDR = ("# SPDX-License-Identifier: GPL-3.0-or-later\n"
       "# Copyright (C) 2026 Viacheslav Manin\n\n")
for p in pathlib.Path(".").glob("*.py"):
    txt = p.read_text(encoding="utf-8")
    if "SPDX-License-Identifier" in txt:
        continue
    m = re.match(r"(#!.*\n|#.*coding[:=].*\n)*", txt)
    cut = m.end() if m else 0
    p.write_text(txt[:cut] + HDR + txt[cut:], encoding="utf-8")
    
```
